# Which team won the most matches?

```python
match_win = df['match_winner'].value_counts()
```

```python
match_win
```

```python
sns.barplot(y = match_win.index, x =match_win.values, palette='rainbow')
plt.title("Most Match Wins by a Team")
```

# TOSS DECISION

```python
sns.countplot(x ='toss_decision', data = df, palette = 'rainbow')
plt.title("Toss Decision Trends")
```

# TOSS WINNER VS MATCH WINNER

```python
count = df[df['toss_winner'] ==df['match_winner']]['match_id'].count()
percentage = (count*100)/df.shape[0]
percentage.round(2)
```

## 3. Top 10 Best Bowling Figures

```python
df['highest_wicket'] =df['best_bowling_figure'].apply(lambda x : x.split('--')[0])
```

```python
df['highest_wicket'] = df['highest_wicket'].astype(int)
```

```python
df['highest_wicket']
```

```python
top_bowling = df.groupby('best_bowling')['highest_wicket'].sum().sort_values(ascending=False).head(10)
```

```python
top_bowling
```

```python
top_bowling.plot(kind = 'barh')
```
